```python
import numpy as np


def diff_quotient(f, x, h):
 return (f(x + h) - f(x)) / h


f = np.sin
x = 1.0
true = np.cos(x)
hs = 10.0 ** (-np.arange(1, 17)) # 1e-1 ... 1e-16
approx = np.array([diff_quotient(f, x, h) for h in hs])


err = np.abs(approx - true)
# Show a small readable table
print(" h approx |error|")
for h,a,e in zip(hs, approx, err):
 print(f"{h: .1e} {a: .16f} {e: .3e}")
```
